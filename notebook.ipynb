### 🛠️ Coding Challenge 2 (Tricky Edge Case): Truthiness Audit

**Scenario:** In Python, many objects are Falsy (like `0`, `[]`, `None`). But sometimes we need to do strict type/value auditing.

**Task:** Write a function `evaluate_truthiness(items: list) -> dict` that audits a list of values. It should count the truthy and falsy values and collect the type names of all falsy elements.

**Rules:**
- Return a dictionary with the exact keys: `'total_elements'`, `'truthy_count'`, `'falsy_count'`, and `'falsy_types'`.
- `'falsy_types'` must be a **set of strings** containing the `__name__` of the types of all falsy values (e.g. `"bool"`, `"int"`, `"NoneType"`, `"list"`).
- **Watch out for booleans vs integers!** Remember that `False` is an instance of `int` in Python (`isinstance(False, int)` is `True`). Make sure your logic does not misclassify `False` as just a plain `int` for type names.

```python
def evaluate_truthiness(items: list) -> dict:
    # TODO: Implement truthiness evaluation
    res = {'total_elements' : 0, 'truthy_count' : 0 , 'falsy_count' : 0, 'falsy_types' : set()}
    for i in items:
        res['total_elements'] = res.get('total_elements') + 1
        if i:
            res['truthy_count'] = res.get('truthy_count') + 1
        else:
            res["falsy_count"] = res.get("falsy_count") + 1
            res['falsy_types'].add(type(i).__name__)
    return res
        

# --- Verification Code ---
try:
    test_list = [True, False, 0, 0.0, 0j, "", [], {}, (), set(), None, 42, "hello"]
    result = evaluate_truthiness(test_list)
    
    assert result['total_elements'] == 13, f"Expected 13 elements, got {result['total_elements']}"
    assert result['truthy_count'] == 3, f"Expected 3 truthy elements (True, 42, 'hello'), got {result['truthy_count']}"
    assert result['falsy_count'] == 10, f"Expected 10 falsy elements, got {result['falsy_count']}"
    
    expected_types = {'bool', 'int', 'float', 'complex', 'str', 'list', 'dict', 'tuple', 'set', 'NoneType'}
    assert result['falsy_types'] == expected_types, f"Expected types {expected_types}, got {result['falsy_types']}"
    print("🎉 Challenge 2: Success! All assertions passed.")
except AssertionError as e:
    print("❌ Assertion failed:", e)
```

### 🛠️ Coding Challenge 1 (Beginner): Parse & Format User Data

**Scenario:** You are processing a comma-separated CSV log of user data. The data is extremely dirty, with trailing spaces, newlines, and inconsistent casing.

**Task:** Write a function `format_user_data(user_str: str) -> str` that parses a dirty string containing `"username, age, profession"` and returns a clean, formatted f-string.

**Rules:**
- Trim all leading and trailing whitespace from the string and its individual parts.
- Convert the username (which contains underscores, e.g. `john_doe`) into a spaced, title-cased name (e.g. `John Doe`).
- Age must be a clean string. If it is empty or missing, default to `"Unknown"`.
- Profession must be title-cased. If it is empty or missing, default to `"Unknown"`.
- The output format must match exactly: `"User: <Clean Name> | Age: <Clean Age> | Profession: <Clean Profession>"`.

```python
def format_user_data(user_str: str) -> str:
    # TODO: Clean and format user data
     parts = user_str.split(',')
     
     user_raw = parts[0].strip() if len(parts) > 0 else ""
     age_raw = parts[1].strip()if len(parts) > 1 else ""
     pro_raw = parts[2].strip()if len(parts) > 2 else ""
     
     if user_raw:
         name = user_raw.replace('-'," ").title()
     else:
         name = "Unknown" 
     
     age = age_raw if age_raw else "Unknown"
     pro = pro_raw.title() if pro_raw else "Unknown"
     
     return f"User: {name} | Age: {age} | Profession: {pro}"    
    
# print(format_user_data("  john_doe , 28 ,  engineer\n"))
# --- Verification Code ---
try:
    assert format_user_data("  john_doe , 28 ,  engineer\n") == "User: John Doe | Age: 28 | Profession: Engineer"
    assert format_user_data("ALICE_SMITH,32,") == "User: Alice Smith | Age: 32 | Profession: Unknown"
    assert format_user_data("  bob_jones , , writer ") == "User: Bob Jones | Age: Unknown | Profession: Writer"
    assert format_user_data(" , , ") == "User: Unknown | Age: Unknown | Profession: Unknown"
    print("🎉 Challenge 1: Success! All assertions passed.")
except AssertionError as e:
    print("❌ Assertion failed:", e)
```
